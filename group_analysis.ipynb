# Group Analysis - Marine Species Taxonomic Analysis

Ce notebook effectue une analyse de groupe des espèces marines basée sur leur taxonomie (Class, Family, Order, etc.) en utilisant le même principe que le notebook interactif mais appliqué aux groupes taxonomiques plutôt qu'aux espèces individuelles.

## Objectifs :
1. Merger les données `marine_species_expanded.csv` avec `AmP_species_list.csv` pour obtenir les informations taxonomiques
2. Échantillonner 500,000 observations
3. Créer une grille mondiale 100x100
4. Agréger les paramètres par cellule et par groupe taxonomique
5. Analyser les patterns par groupes taxonomiques

## 3. Merge des Datasets

```python
# Merge des datasets - Version corrigée
if df_marine is not None and df_taxonomy is not None:
    print("🔗 Merge des datasets - Version corrigée...")
    
    # Identifier les colonnes pour le merge
    species_col_marine = 'species'
    genus_col_marine = 'genus'
    taxonomy_col = 'ScientificName'
    
    print(f"📌 Colonnes identifiées pour merge (marine): {genus_col_marine} + {species_col_marine}")
    print(f"📌 Colonne pour merge (taxonomique): {taxonomy_col}")
    
    # Créer une copie des datasets pour le nettoyage
    df_marine_clean = df_marine.copy()
    df_taxonomy_clean = df_taxonomy.copy()
    
    # ÉTAPE 1: Créer le nom scientifique complet dans le dataset marine
    print("🧬 Création du nom scientifique complet (genus + species)...")
    df_marine_clean['scientific_name_full'] = (
        df_marine_clean[genus_col_marine].astype(str).str.strip() + ' ' + 
        df_marine_clean[species_col_marine].astype(str).str.strip()
    )
    
    # Nettoyage des noms scientifiques
    def clean_scientific_name(name):
        if pd.isna(name) or name == 'nan nan':
            return None
        return str(name).strip().lower().replace('_', ' ')
    
    df_marine_clean['scientific_name_clean'] = df_marine_clean['scientific_name_full'].apply(clean_scientific_name)
    df_taxonomy_clean['scientific_name_clean'] = df_taxonomy_clean[taxonomy_col].apply(clean_scientific_name)
    
    # Afficher quelques exemples
    print("\n🔍 Exemples de noms scientifiques créés:")
    marine_sample = df_marine_clean[['genus', 'species', 'scientific_name_full', 'scientific_name_clean']].dropna().head()
    for _, row in marine_sample.iterrows():
        print(f"  {row['genus']} + {row['species']} → '{row['scientific_name_full']}' → '{row['scientific_name_clean']}'")
    
    print(f"\n🧬 Exemples du dataset taxonomique:")
    taxonomy_sample = df_taxonomy_clean[['ScientificName', 'scientific_name_clean']].dropna().head()
    for _, row in taxonomy_sample.iterrows():
        print(f"  '{row['ScientificName']}' → '{row['scientific_name_clean']}'")
    
    # ÉTAPE 2: Effectuer le merge
    print(f"\n🔗 Merge sur les noms scientifiques nettoyés...")
    df_merged = df_marine_clean.merge(
        df_taxonomy_clean,
        left_on='scientific_name_clean',
        right_on='scientific_name_clean',
        how='left',
        suffixes=('', '_taxonomy')
    )
    
    # Calculer les statistiques de merge
    total_observations = len(df_merged)
    with_taxonomy = df_merged['Phylum'].notna().sum()
    match_rate = (with_taxonomy / total_observations) * 100
    
    print(f"✅ Merge terminé: {df_merged.shape}")
    print(f"📊 Observations avec données taxonomiques: {with_taxonomy:,}")
    print(f"📊 Taux de match final: {match_rate:.1f}%")
    
    if match_rate > 0:
        print(f"🎉 Succès ! {match_rate:.1f}% des observations ont été enrichies avec des données taxonomiques")
        
        # Afficher la distribution par classe taxonomique
        if 'Class' in df_merged.columns:
            class_counts = df_merged['Class'].value_counts().head(10)
            print(f"\n📊 Top 10 des classes taxonomiques:")
            for class_name, count in class_counts.items():
                print(f"  {class_name}: {count:,} observations")
        
        strategy_used = "TAXONOMIQUE"
        print(f"✅ Stratégie utilisée pour l'analyse: {strategy_used}")
        
    else:
        print(f"⚠️ Taux de match très faible ({match_rate:.1f}%) - Diagnostic nécessaire")
        
        # Échantillons pour diagnostic
        print(f"\n🔍 Échantillons de noms pour diagnostic:")
        marine_names = df_marine_clean['scientific_name_clean'].dropna().unique()[:5]
        taxonomy_names = df_taxonomy_clean['scientific_name_clean'].dropna().unique()[:5]
        
        print(f"Marine (premiers 5):")
        for name in marine_names:
            print(f"  '{name}'")
        
        print(f"\nTaxonomique (premiers 5):")
        for name in taxonomy_names:
            print(f"  '{name}'")
        
        # Vérifier les correspondances directes
        marine_set = set(df_marine_clean['scientific_name_clean'].dropna())
        taxonomy_set = set(df_taxonomy_clean['scientific_name_clean'].dropna())
        direct_matches = marine_set.intersection(taxonomy_set)
        print(f"\n📈 Correspondances directes trouvées: {len(direct_matches)}")
        if len(direct_matches) > 0:
            print(f"Exemples: {list(direct_matches)[:5]}")
else:
    print("❌ Impossible de merger - datasets non chargés")
    df_merged = None
```

## 4. Échantillonnage de 500,000 Observations

```python
# Échantillonnage stratifié de 500,000 observations
if 'df_merged' in locals():
    print("🎯 Échantillonnage de 500,000 observations...")
    
    # Vérifier d'abord s'il y a des données taxonomiques
    df_with_taxonomy = df_merged[df_merged['Class'].notna()].copy()
    print(f"📊 Observations avec taxonomie: {len(df_with_taxonomy)}")
    
    if len(df_with_taxonomy) == 0:
        print("⚠️ Aucune donnée taxonomique trouvée !")
        print("🔄 Stratégie alternative: Utilisation de groupes basés sur les données existantes...")
        
        # Créer des groupes artificiels basés sur d'autres colonnes
        # Par exemple, par région géographique ou autres caractéristiques
        
        # Identifier les colonnes disponibles pour créer des groupes
        available_cols = df_merged.columns.tolist()
        print(f"📋 Colonnes disponibles: {available_cols[:10]}...")
        
        # Essayer de créer des groupes géographiques
        if any('lat' in col.lower() for col in available_cols) and any('lon' in col.lower() for col in available_cols):
            print("🌍 Création de groupes géographiques...")
            
            lat_col = next((col for col in available_cols if 'lat' in col.lower()), None)
            lon_col = next((col for col in available_cols if 'lon' in col.lower()), None)
            
            if lat_col and lon_col:
                # Créer des zones géographiques
                df_merged['geo_zone'] = 'Unknown'
                
                # Zones basées sur la latitude
                df_merged.loc[df_merged[lat_col] > 60, 'geo_zone'] = 'Arctic'
                df_merged.loc[(df_merged[lat_col] <= 60) & (df_merged[lat_col] > 23.5), 'geo_zone'] = 'Northern_Temperate'
                df_merged.loc[(df_merged[lat_col] <= 23.5) & (df_merged[lat_col] > -23.5), 'geo_zone'] = 'Tropical'
                df_merged.loc[(df_merged[lat_col] <= -23.5) & (df_merged[lat_col] > -60), 'geo_zone'] = 'Southern_Temperate'
                df_merged.loc[df_merged[lat_col] <= -60, 'geo_zone'] = 'Antarctic'
                
                # Utiliser geo_zone comme groupe de substitution
                selected_group_column = 'geo_zone'
                df_with_groups = df_merged[df_merged[selected_group_column] != 'Unknown'].copy()
                
                print(f"✅ Groupes géographiques créés:")
                print(df_with_groups[selected_group_column].value_counts())
                
            else:
                print("❌ Impossible de créer des groupes géographiques")
                df_with_groups = df_merged.copy()
                selected_group_column = None
        else:
            print("❌ Données insuffisantes pour créer des groupes")
            df_with_groups = df_merged.copy()
            selected_group_column = None
            
    else:
        # Utilisation normale avec données taxonomiques
        df_with_groups = df_with_taxonomy.copy()
        selected_group_column = 'Class'
    
    # Échantillonnage
    sample_size = min(500000, len(df_with_groups))
    
    if len(df_with_groups) > sample_size and selected_group_column:
        # Échantillonnage stratifié par groupe
        df_sample = df_with_groups.groupby(selected_group_column, group_keys=False).apply(
            lambda x: x.sample(n=min(len(x), max(1, int(sample_size * len(x) / len(df_with_groups)))), 
                              random_state=42)
        ).reset_index(drop=True)
        
        # Compléter l'échantillon si nécessaire
        if len(df_sample) < sample_size:
            remaining = sample_size - len(df_sample)
            remaining_data = df_with_groups[~df_with_groups.index.isin(df_sample.index)]
            if len(remaining_data) > 0:
                additional_sample = remaining_data.sample(n=min(remaining, len(remaining_data)), random_state=42)
                df_sample = pd.concat([df_sample, additional_sample], ignore_index=True)
    else:
        df_sample = df_with_groups.copy()
    
    print(f"✅ Échantillon final: {len(df_sample)} observations")
    
    if selected_group_column and selected_group_column in df_sample.columns:
        print(f"🧬 Distribution par {selected_group_column}:")
        print(df_sample[selected_group_column].value_counts().head(10))
    else:
        print("⚠️ Pas de groupement disponible pour l'échantillonnage")
    
    # Libération de mémoire
    if 'df_merged' in locals():
        del df_merged
    if 'df_with_taxonomy' in locals():
        del df_with_taxonomy
    if 'df_with_groups' in locals():
        del df_with_groups
    gc.collect()
    
else:
    print("❌ Données mergées non disponibles")
```

## 5. Création de la Grille Mondiale 100x100

```python
# Configuration de la grille mondiale 100x100
if 'df_sample' in locals():
    print("🌍 Création de la grille mondiale 100x100...")
    
    # Paramètres de la grille
    grid_size = 100
    lat_min, lat_max = -90, 90
    lon_min, lon_max = -180, 180
    
    # Calcul des intervalles
    lat_step = (lat_max - lat_min) / grid_size
    lon_step = (lon_max - lon_min) / grid_size
    
    print(f"📐 Taille de cellule: {lat_step:.2f}° latitude × {lon_step:.2f}° longitude")
    
    # Fonction pour assigner les cellules de grille
    def assign_grid_cell(lat, lon):
        """Assigne une cellule de grille basée sur lat/lon"""
        # Éviter les valeurs exactement sur les bords
        lat = np.clip(lat, lat_min + 1e-10, lat_max - 1e-10)
        lon = np.clip(lon, lon_min + 1e-10, lon_max - 1e-10)
        
        lat_idx = int((lat - lat_min) / lat_step)
        lon_idx = int((lon - lon_min) / lon_step)
        
        # S'assurer que les indices sont dans les limites
        lat_idx = np.clip(lat_idx, 0, grid_size - 1)
        lon_idx = np.clip(lon_idx, 0, grid_size - 1)
        
        return lat_idx, lon_idx
    
    # Identifier les colonnes de coordonnées
    lat_col = None
    lon_col = None
    
    for col in df_sample.columns:
        if 'lat' in col.lower() and lat_col is None:
            lat_col = col
        if 'lon' in col.lower() and lon_col is None:
            lon_col = col
    
    print(f"📍 Colonnes identifiées - Latitude: {lat_col}, Longitude: {lon_col}")
    
    if lat_col and lon_col:
        # Filtrer les données avec coordonnées valides
        df_sample_coords = df_sample[
            (df_sample[lat_col].notna()) & 
            (df_sample[lon_col].notna()) &
            (df_sample[lat_col] >= lat_min) & 
            (df_sample[lat_col] <= lat_max) &
            (df_sample[lon_col] >= lon_min) & 
            (df_sample[lon_col] <= lon_max)
        ].copy()
        
        print(f"📊 Observations avec coordonnées valides: {len(df_sample_coords)}")
        
        # Assigner les cellules de grille
        grid_assignments = df_sample_coords.apply(
            lambda row: assign_grid_cell(row[lat_col], row[lon_col]), axis=1
        )
        
        df_sample_coords['grid_lat'] = [x[0] for x in grid_assignments]
        df_sample_coords['grid_lon'] = [x[1] for x in grid_assignments]
        
        # Calcul des coordonnées du centre des cellules
        df_sample_coords['center_lat'] = lat_min + (df_sample_coords['grid_lat'] + 0.5) * lat_step
        df_sample_coords['center_lon'] = lon_min + (df_sample_coords['grid_lon'] + 0.5) * lon_step
        
        print(f"✅ Grille assignée - Cellules occupées: {df_sample_coords[['grid_lat', 'grid_lon']].drop_duplicates().shape[0]}")
        
    else:
        print("❌ Colonnes de coordonnées non trouvées")
        df_sample_coords = None
else:
    print("❌ Échantillon non disponible")
```

## 6. Choix du Groupe Taxonomique pour l'Analyse

```python
# Analyse des groupes taxonomiques disponibles
if 'df_sample_coords' in locals() and df_sample_coords is not None:
    print("🧬 Analyse des groupes taxonomiques disponibles...")
    
    # Analyse par niveau taxonomique
    taxonomic_levels = ['Phylum', 'Class', 'Order', 'Family']
    
    for level in taxonomic_levels:
        if level in df_sample_coords.columns:
            level_counts = df_sample_coords[level].value_counts()
            print(f"\n📊 {level} (Top 10):")
            print(level_counts.head(10))
            print(f"Total unique {level}: {level_counts.shape[0]}")
    
    # Recommandation pour le choix du groupe
    print("\n💡 Recommandations pour l'analyse de groupe:")
    
    # Choisir automatiquement la Classe comme niveau d'analyse principal
    if 'Class' in df_sample_coords.columns:
        selected_group_column = 'Class'
        print(f"🎯 Niveau sélectionné: {selected_group_column}")
        
        # Filtrer les classes avec assez d'observations
        min_observations = 100  # Minimum d'observations par groupe
        class_counts = df_sample_coords[selected_group_column].value_counts()
        valid_classes = class_counts[class_counts >= min_observations].index.tolist()
        
        print(f"📈 Classes avec >= {min_observations} observations: {len(valid_classes)}")
        print("🐟 Classes sélectionnées:")
        for cls in valid_classes[:10]:  # Afficher les 10 premières
            print(f"  - {cls}: {class_counts[cls]} observations")
        
        # Filtrer le dataset
        df_analysis = df_sample_coords[
            df_sample_coords[selected_group_column].isin(valid_classes)
        ].copy()
        
        print(f"\n✅ Dataset final pour analyse: {len(df_analysis)} observations")
        print(f"🌍 Cellules de grille représentées: {df_analysis[['grid_lat', 'grid_lon']].drop_duplicates().shape[0]}")
        
    else:
        print("❌ Colonne 'Class' non disponible")
        selected_group_column = None
        df_analysis = None
else:
    print("❌ Données avec coordonnées non disponibles")
```

## 7. Agrégation des Données par Cellule et par Groupe

```python
# Agrégation des données par cellule et par groupe
if 'df_analysis' in locals() and df_analysis is not None:
    print("📊 Agrégation des données par cellule et par groupe...")
    
    # Créer la colonne group_analysis qui référence le niveau taxonomique choisi
    if 'Class' in df_analysis.columns:
        df_analysis['group_analysis'] = df_analysis['Class']
        print(f"🧬 Groupe d'analyse: Classe taxonomique")
    else:
        print("❌ Colonne 'Class' non disponible pour l'analyse")
        df_analysis = None
    
    if df_analysis is not None:
        # Colonnes pour l'agrégation (paramètres environnementaux)
        agg_columns = [col for col in df_analysis.columns if 'param' in col.lower()]
        print(f"📋 Colonnes pour agrégation: {len(agg_columns)}")
        print(f"📋 Colonnes: {agg_columns[:10]}{'...' if len(agg_columns) > 10 else ''}")
        
        # Groupement par cellule de grille et groupe taxonomique
        grouping_cols = ['grid_lat', 'grid_lon', 'group_analysis']
        
        # Créer une colonne pour compter les observations
        df_analysis['observation_count'] = 1
        
        # Dictionnaire d'agrégation
        agg_dict = {}
        for col in agg_columns:
            if col in df_analysis.columns:
                agg_dict[col] = ['mean', 'std', 'min', 'max', 'count']
        
        # Ajouter le compte d'observations
        agg_dict['observation_count'] = 'sum'  # Somme pour compter les observations
        
        print("🔄 Calcul des agrégations...")
        df_grid_groups = df_analysis.groupby(grouping_cols).agg(agg_dict).reset_index()
        
        # Aplatir les colonnes multi-niveau
        new_columns = []
        for col in df_grid_groups.columns:
            if isinstance(col, tuple):
                if col[1] == '':
                    new_columns.append(col[0])
                else:
                    new_columns.append(f"{col[0]}_{col[1]}")
            else:
                new_columns.append(col)
        
        df_grid_groups.columns = new_columns
        
        print(f"✅ Agrégation terminée: {df_grid_groups.shape}")
        print(f"📊 Combinaisons cellule-groupe: {len(df_grid_groups)}")
        
        # Afficher quelques statistiques
        print(f"\n📈 Statistiques de l'agrégation:")
        print(f"  - Cellules de grille uniques: {df_grid_groups[['grid_lat', 'grid_lon']].drop_duplicates().shape[0]}")
        print(f"  - Groupes taxonomiques: {df_grid_groups['group_analysis'].nunique()}")
        print(f"  - Observations par cellule-groupe (moyenne): {df_grid_groups['observation_count_sum'].mean():.1f}")
        print(f"  - Observations par cellule-groupe (médiane): {df_grid_groups['observation_count_sum'].median():.1f}")
        
        # Distribution des observations par groupe
        group_stats = df_grid_groups.groupby('group_analysis')['observation_count_sum'].agg(['sum', 'count', 'mean']).sort_values('sum', ascending=False)
        print(f"\n🧬 Distribution par groupe taxonomique:")
        for group_name, stats in group_stats.head(10).iterrows():
            print(f"  - {group_name}: {stats['sum']:,} obs dans {stats['count']} cellules (moy: {stats['mean']:.1f} obs/cellule)")
        
        print(f"\n✅ Dataset agrégé prêt pour l'analyse et la visualisation")
    else:
        df_grid_groups = None
else:
    print("❌ Dataset d'analyse non disponible")
    df_grid_groups = None
```

```python
# Préparation des données pour machine learning
print("🤖 Préparation des données pour machine learning...")

if 'df_grid_groups' in locals() and df_grid_groups is not None:
    # Créer le dataset ML avec probabilité maximale par cellule pour chaque classe
    print("📊 Création du dataset ML avec probabilité maximale par cellule...")
    
    # Obtenir les coordonnées de toutes les cellules uniques
    unique_cells = df_grid_groups[['grid_lat', 'grid_lon']].drop_duplicates().copy()
    print(f"🌍 Cellules uniques dans la grille: {len(unique_cells)}")
    
    # Variables environnementales (moyennes)
    env_columns = [col for col in df_grid_groups.columns if col.endswith('_mean_mean')]
    print(f"🌡️ Variables environnementales: {len(env_columns)}")
    
    # Pour chaque cellule, calculer la probabilité maximale pour chaque classe
    classes = df_grid_groups['group_analysis'].unique()
    print(f"🧬 Classes taxonomiques: {len(classes)}")
    
    # Créer le dataset ML final
    ml_data_list = []
    
    for _, cell in unique_cells.iterrows():
        cell_data = df_grid_groups[
            (df_grid_groups['grid_lat'] == cell['grid_lat']) & 
            (df_grid_groups['grid_lon'] == cell['grid_lon'])
        ]
        
        if len(cell_data) > 0:
            # Coordonnées de la cellule
            row = {
                'grid_lat': cell['grid_lat'],
                'grid_lon': cell['grid_lon'],
                'center_lat': cell['grid_lat'] * 1.8 - 90 + 0.9,
                'center_lon': cell['grid_lon'] * 3.6 - 180 + 1.8
            }
            
            # Variables environnementales (moyenne des moyennes pour la cellule)
            for env_col in env_columns:
                if env_col in cell_data.columns:
                    row[env_col.replace('_mean_mean', '')] = cell_data[env_col].mean()
            
            # Probabilité maximale pour chaque classe (normalisée)
            total_obs = cell_data['observation_count_sum'].sum()
            for class_name in classes:
                class_obs = cell_data[cell_data['group_analysis'] == class_name]['observation_count_sum'].sum()
                row[f'prob_{class_name.lower().replace(" ", "_")}'] = class_obs / total_obs if total_obs > 0 else 0
            
            # Classe dominante
            max_prob_class = cell_data.loc[cell_data['observation_count_sum'].idxmax(), 'group_analysis']
            row['dominant_class'] = max_prob_class
            row['max_probability'] = cell_data['observation_count_sum'].max() / total_obs if total_obs > 0 else 0
            row['total_observations'] = total_obs
            
            ml_data_list.append(row)
    
    # Créer le DataFrame ML
    df_ml = pd.DataFrame(ml_data_list)
    print(f"✅ Dataset ML créé: {df_ml.shape}")
    
    # Afficher les statistiques
    print(f"\n📊 Statistiques du dataset ML:")
    print(f"  - Cellules: {len(df_ml)}")
    print(f"  - Variables environnementales: {len(env_columns)}")
    print(f"  - Classes taxonomiques: {len(classes)}")
    print(f"  - Observations totales: {df_ml['total_observations'].sum():,}")
    
    # Distribution des classes dominantes
    print(f"\n🧬 Distribution des classes dominantes:")
    dominant_dist = df_ml['dominant_class'].value_counts().head()
    for class_name, count in dominant_dist.items():
        percentage = (count / len(df_ml)) * 100
        print(f"  - {class_name}: {count} cellules ({percentage:.1f}%)")
    
    print("\n✅ Données prêtes pour l'entraînement ML")
    
else:
    print("❌ Données agrégées non disponibles")
    df_ml = None
```

```python
# Entraînement du modèle Random Forest
if 'df_ml' in locals() and df_ml is not None:
    print("🚀 Entraînement du modèle Random Forest...")
    
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import classification_report, confusion_matrix
    from sklearn.preprocessing import StandardScaler
    import joblib
    
    # Préparer les features (variables environnementales)
    feature_columns = [col for col in df_ml.columns if col in [c.replace('_mean_mean', '') for c in env_columns]]
    X = df_ml[feature_columns].copy()
    y = df_ml['dominant_class'].copy()
    
    print(f"📊 Features: {len(feature_columns)} variables")
    print(f"🎯 Target: classes taxonomiques ({y.nunique()} classes)")
    print(f"📝 Features utilisées: {feature_columns[:5]}{'...' if len(feature_columns) > 5 else ''}")
    
    # Gérer les valeurs manquantes
    X = X.fillna(X.mean())
    
    # CORRECTION : Filtrer les classes avec trop peu d'observations pour la stratification
    class_counts = y.value_counts()
    print(f"\n🔍 Distribution des classes avant filtrage:")
    print(class_counts.head(10))
    
    # Garder seulement les classes avec au moins 2 observations (minimum pour stratification)
    valid_classes = class_counts[class_counts >= 2].index
    mask = y.isin(valid_classes)
    
    X_filtered = X[mask].copy()
    y_filtered = y[mask].copy()
    
    print(f"\n✂️ Filtrage des classes:")
    print(f"  - Classes avant filtrage: {y.nunique()}")
    print(f"  - Classes après filtrage: {y_filtered.nunique()}")
    print(f"  - Observations avant filtrage: {len(X)}")
    print(f"  - Observations après filtrage: {len(X_filtered)}")
    
    if len(y_filtered.unique()) < 2:
        print("❌ Pas assez de classes pour l'entraînement ML")
    else:
        # Division train/test avec les données filtrées
        X_train, X_test, y_train, y_test = train_test_split(
            X_filtered, y_filtered, test_size=0.2, random_state=42, stratify=y_filtered
        )
        
        print(f"\n✂️ Division des données:")
        print(f"  - Training: {len(X_train)} cellules")
        print(f"  - Test: {len(X_test)} cellules")
        
        # Standardisation
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)
        
        # Entraînement du modèle Random Forest
        print("🌲 Entraînement Random Forest...")
        rf_model = RandomForestClassifier(
            n_estimators=100,
            max_depth=10,
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=42,
            n_jobs=-1
        )
        
        rf_model.fit(X_train_scaled, y_train)
        
        # Prédictions
        y_pred = rf_model.predict(X_test_scaled)
        y_pred_proba = rf_model.predict_proba(X_test_scaled)
        
        # Évaluation
        print(f"\n📈 Évaluation du modèle:")
        print(f"  - Accuracy: {rf_model.score(X_test_scaled, y_test):.3f}")
        
        # Importance des features
        feature_importance = pd.DataFrame({
            'feature': feature_columns,
            'importance': rf_model.feature_importances_
        }).sort_values('importance', ascending=False)
        
        print(f"\n🔝 Top 5 variables les plus importantes:")
        for _, row in feature_importance.head().iterrows():
            print(f"  - {row['feature']}: {row['importance']:.3f}")
        
        # Sauvegarder le modèle
        joblib.dump(rf_model, 'taxonomic_classifier_model.pkl')
        joblib.dump(scaler, 'taxonomic_scaler.pkl')
        print(f"\n💾 Modèle sauvegardé: taxonomic_classifier_model.pkl")
        
        print("✅ Entraînement terminé avec succès")
        
else:
    print("❌ Dataset ML non disponible pour l'entraînement")
```

```python
# Prédictions sur l'ensemble du dataset et évaluation
if 'rf_model' in locals() and 'df_ml' in locals():
    print("🔮 Génération des prédictions...")
    
    # Prédictions sur tout le dataset
    X_all = df_ml[feature_columns].fillna(df_ml[feature_columns].mean())
    X_all_scaled = scaler.transform(X_all)
    
    # Prédictions de classe et probabilités
    pred_classes = rf_model.predict(X_all_scaled)
    pred_probabilities = rf_model.predict_proba(X_all_scaled)
    
    # Ajouter les prédictions au dataset
    df_ml_pred = df_ml.copy()
    df_ml_pred['predicted_class'] = pred_classes
    df_ml_pred['prediction_confidence'] = pred_probabilities.max(axis=1)
    
    # Ajouter les probabilités pour chaque classe
    class_names = rf_model.classes_
    for i, class_name in enumerate(class_names):
        df_ml_pred[f'pred_prob_{class_name.lower().replace(" ", "_")}'] = pred_probabilities[:, i]
    
    print(f"✅ Prédictions générées pour {len(df_ml_pred)} cellules")
    
    # Évaluation détaillée
    print(f"\n📊 Évaluation détaillée:")
    accuracy = (df_ml_pred['dominant_class'] == df_ml_pred['predicted_class']).mean()
    print(f"  - Accuracy globale: {accuracy:.3f}")
    
    # Accuracy par classe
    print(f"\n🧬 Accuracy par classe:")
    for class_name in class_names[:5]:  # Top 5 classes
        class_mask = df_ml_pred['dominant_class'] == class_name
        if class_mask.sum() > 0:
            class_accuracy = (
                df_ml_pred.loc[class_mask, 'dominant_class'] == 
                df_ml_pred.loc[class_mask, 'predicted_class']
            ).mean()
            print(f"  - {class_name}: {class_accuracy:.3f} ({class_mask.sum()} cellules)")
    
    # Distribution des confidences
    print(f"\n🎯 Distribution des confidences de prédiction:")
    confidence_stats = df_ml_pred['prediction_confidence'].describe()
    print(f"  - Moyenne: {confidence_stats['mean']:.3f}")
    print(f"  - Médiane: {confidence_stats['50%']:.3f}")
    print(f"  - Min: {confidence_stats['min']:.3f}")
    print(f"  - Max: {confidence_stats['max']:.3f}")
    
    # Cellules avec haute confiance
    high_confidence = df_ml_pred[df_ml_pred['prediction_confidence'] > 0.8]
    print(f"  - Cellules avec confiance > 80%: {len(high_confidence)} ({len(high_confidence)/len(df_ml_pred)*100:.1f}%)")
    
    # Sauvegarder les résultats
    df_ml_pred.to_csv('ml_predictions_complete.csv', index=False)
    
    # Créer un résumé des prédictions
    summary_data = []
    for class_name in class_names:
        predicted_cells = (df_ml_pred['predicted_class'] == class_name).sum()
        actual_cells = (df_ml_pred['dominant_class'] == class_name).sum()
        avg_confidence = df_ml_pred[df_ml_pred['predicted_class'] == class_name]['prediction_confidence'].mean()
        
        summary_data.append({
            'class': class_name,
            'actual_cells': actual_cells,
            'predicted_cells': predicted_cells,
            'avg_confidence': avg_confidence if not pd.isna(avg_confidence) else 0
        })
    
    df_summary = pd.DataFrame(summary_data).sort_values('predicted_cells', ascending=False)
    df_summary.to_csv('ml_predictions_summary.csv', index=False)
    
    print(f"\n💾 Résultats sauvegardés:")
    print(f"  - ml_predictions_complete.csv (prédictions complètes)")
    print(f"  - ml_predictions_summary.csv (résumé par classe)")
    
    print("\n✅ Prédictions et évaluation terminées")
    
else:
    print("❌ Modèle ou données non disponibles pour les prédictions")
```

```python
# ==============================================================================
# PRÉDICTION MACHINE LEARNING PAR CATÉGORIE TAXONOMIQUE
# ==============================================================================
# Code général pour prédire la distribution d'une catégorie taxonomique
# Changez simplement la variable 'category_name' pour analyser différentes catégories

# ⚙️ CONFIGURATION - Changez cette ligne pour analyser différentes catégories
category_name = "Mammalia"  # Peut être changé pour: "Actinopterygii", "Aves", "Reptilia", etc.

print(f"🧬 Analyse ML pour la catégorie: {category_name}")
print("="*60)

if 'df_grid_groups' in locals() and df_grid_groups is not None:
    # Vérifier si la catégorie existe dans les données
    available_categories = df_grid_groups['group_analysis'].unique()
    print(f"📋 Catégories disponibles: {len(available_categories)}")
    print(f"🔍 Top 10 catégories: {available_categories[:10].tolist()}")
    
    if category_name not in available_categories:
        print(f"❌ Catégorie '{category_name}' non trouvée dans les données")
        print(f"💡 Catégories similaires disponibles:")
        similar_categories = [cat for cat in available_categories if category_name.lower() in cat.lower()]
        if similar_categories:
            for cat in similar_categories[:5]:
                print(f"   - {cat}")
        else:
            print(f"   Aucune catégorie similaire trouvée")
            print(f"   Essayez une de ces catégories: {available_categories[:5].tolist()}")
    else:
        print(f"✅ Catégorie '{category_name}' trouvée dans les données")
        
        # Filtrer les données pour la catégorie sélectionnée
        category_data = df_grid_groups[df_grid_groups['group_analysis'] == category_name].copy()
        print(f"📊 Données pour {category_name}:")
        print(f"   - Cellules avec observations: {len(category_data)}")
        print(f"   - Observations totales: {category_data['observation_count_sum'].sum():,}")
        print(f"   - Étendue géographique:")
        
        # Calculer l'étendue géographique
        lat_coords = category_data['grid_lat'] * 1.8 - 90 + 0.9
        lon_coords = category_data['grid_lon'] * 3.6 - 180 + 1.8
        print(f"     Latitude: {lat_coords.min():.1f}° à {lat_coords.max():.1f}°")
        print(f"     Longitude: {lon_coords.min():.1f}° à {lon_coords.max():.1f}°")
        
        # Variables environnementales disponibles
        env_columns = [col for col in df_grid_groups.columns if col.endswith('_mean_mean')]
        print(f"🌡️ Variables environnementales disponibles: {len(env_columns)}")
        
        if len(env_columns) > 0 and len(category_data) > 0:
            print(f"\n🤖 Création du modèle ML pour {category_name}...")
            
            # Préparer les données pour ML
            # Créer une grille complète de toutes les cellules possibles
            all_lat_indices = range(100)  # Grille 100x100
            all_lon_indices = range(100)
            
            ml_prediction_data = []
            
            for lat_idx in all_lat_indices:
                for lon_idx in all_lon_indices:
                    # Coordonnées géographiques réelles
                    center_lat = lat_idx * 1.8 - 90 + 0.9
                    center_lon = lon_idx * 3.6 - 180 + 1.8
                    
                    # Chercher les données environnementales pour cette cellule
                    cell_env_data = df_grid_groups[
                        (df_grid_groups['grid_lat'] == lat_idx) & 
                        (df_grid_groups['grid_lon'] == lon_idx)
                    ]
                    
                    if len(cell_env_data) > 0:
                        # Utiliser les données environnementales moyennes de cette cellule
                        row = {
                            'grid_lat': lat_idx,
                            'grid_lon': lon_idx,
                            'center_lat': center_lat,
                            'center_lon': center_lon
                        }
                        
                        # Variables environnementales
                        for env_col in env_columns:
                            if env_col in cell_env_data.columns:
                                row[env_col.replace('_mean_mean', '')] = cell_env_data[env_col].mean()
                        
                        # Vérifier si la catégorie est présente dans cette cellule
                        category_in_cell = cell_env_data[cell_env_data['group_analysis'] == category_name]
                        if len(category_in_cell) > 0:
                            row['observed_presence'] = 1
                            row['observation_count'] = category_in_cell['observation_count_sum'].sum()
                        else:
                            row['observed_presence'] = 0
                            row['observation_count'] = 0
                        
                        ml_prediction_data.append(row)
            
            # Créer le DataFrame pour les prédictions
            df_prediction = pd.DataFrame(ml_prediction_data)
            print(f"📊 Grille de prédiction créée: {len(df_prediction)} cellules")
            
            if len(df_prediction) > 0:
                # Calculer la probabilité de présence basée sur les variables environnementales
                # Utiliser une approche de similarité environnementale
                
                # Normaliser les variables environnementales
                env_feature_columns = [col for col in df_prediction.columns if col in [c.replace('_mean_mean', '') for c in env_columns]]
                
                if len(env_feature_columns) > 0:
                    print(f"🔬 Variables utilisées pour la prédiction: {env_feature_columns[:5]}{'...' if len(env_feature_columns) > 5 else ''}")
                    
                    # Remplir les valeurs manquantes
                    df_prediction_clean = df_prediction.copy()
                    for col in env_feature_columns:
                        if col in df_prediction_clean.columns:
                            df_prediction_clean[col] = df_prediction_clean[col].fillna(df_prediction_clean[col].mean())
                    
                    # Calculer les conditions environnementales moyennes où la catégorie est présente
                    present_cells = df_prediction_clean[df_prediction_clean['observed_presence'] == 1]
                    
                    if len(present_cells) > 0:
                        print(f"✅ Cellules avec présence observée: {len(present_cells)}")
                        
                        # Calculer les moyennes et écarts-types des conditions optimales
                        optimal_conditions = {}
                        for col in env_feature_columns:
                            if col in present_cells.columns:
                                optimal_conditions[col] = {
                                    'mean': present_cells[col].mean(),
                                    'std': present_cells[col].std() if present_cells[col].std() > 0 else 1.0
                                }
                        
                        # Calculer la probabilité de présence pour toutes les cellules
                        probabilities = []
                        
                        for _, row in df_prediction_clean.iterrows():
                            # Calculer la similarité environnementale
                            similarity_scores = []
                            
                            for col in env_feature_columns:
                                if col in optimal_conditions and col in row:
                                    # Distance normalisée par rapport aux conditions optimales
                                    optimal_mean = optimal_conditions[col]['mean']
                                    optimal_std = optimal_conditions[col]['std']
                                    
                                    if not pd.isna(row[col]) and optimal_std > 0:
                                        distance = abs(row[col] - optimal_mean) / optimal_std
                                        # Convertir la distance en score de similarité (0-1)
                                        similarity = np.exp(-distance**2 / 2)
                                        similarity_scores.append(similarity)
                            
                            # Probabilité moyenne basée sur toutes les variables
                            if len(similarity_scores) > 0:
                                probability = np.mean(similarity_scores)
                            else:
                                probability = 0.0
                            
                            probabilities.append(probability)
                        
                        df_prediction_clean['probability'] = probabilities
                        
                        # Classifier en 5 catégories de probabilité
                        def classify_probability(prob):
                            if prob >= 0.8:
                                return "Very Likely"
                            elif prob >= 0.6:
                                return "Likely"
                            elif prob >= 0.4:
                                return "Possible"
                            elif prob >= 0.2:
                                return "Unlikely"
                            else:
                                return "Very Unlikely"
                        
                        df_prediction_clean['probability_class'] = df_prediction_clean['probability'].apply(classify_probability)
                        
                        # Statistiques des prédictions
                        print(f"\n📈 Résultats de la prédiction pour {category_name}:")
                        prob_distribution = df_prediction_clean['probability_class'].value_counts()
                        for prob_class, count in prob_distribution.items():
                            percentage = (count / len(df_prediction_clean)) * 100
                            print(f"   {prob_class}: {count:,} cellules ({percentage:.1f}%)")
                        
                        print(f"\n📊 Statistiques des probabilités:")
                        print(f"   Probabilité moyenne: {df_prediction_clean['probability'].mean():.3f}")
                        print(f"   Probabilité médiane: {df_prediction_clean['probability'].median():.3f}")
                        print(f"   Probabilité max: {df_prediction_clean['probability'].max():.3f}")
                        print(f"   Probabilité min: {df_prediction_clean['probability'].min():.3f}")
                        
                        # Sauvegarder les résultats
                        output_filename = f'ml_prediction_{category_name.lower().replace(" ", "_")}.csv'
                        df_prediction_clean.to_csv(output_filename, index=False)
                        print(f"\n💾 Résultats sauvegardés: {output_filename}")
                        
                        # Créer la visualisation
                        print(f"\n🗺️ Création de la carte de prédiction...")
                        
                        # Définir les couleurs pour chaque classe de probabilité
                        color_map = {
                            "Very Likely": "#d73027",     # Rouge foncé
                            "Likely": "#fc8d59",          # Orange
                            "Possible": "#fee08b",        # Jaune
                            "Unlikely": "#e0f3f8",        # Bleu clair
                            "Very Unlikely": "#91bfdb"    # Bleu
                        }
                        
                        df_prediction_clean['color'] = df_prediction_clean['probability_class'].map(color_map)
                        
                        # Créer la carte avec Plotly
                        fig = px.scatter_geo(
                            df_prediction_clean,
                            lat='center_lat',
                            lon='center_lon',
                            color='probability_class',
                            color_discrete_map=color_map,
                            size='probability',
                            size_max=10,
                            hover_data=['probability', 'observed_presence'],
                            title=f'🧬 Prédiction ML - Distribution de {category_name}',
                            category_orders={"probability_class": ["Very Likely", "Likely", "Possible", "Unlikely", "Very Unlikely"]}
                        )
                        
                        # Configuration de la carte
                        fig.update_geos(
                            projection_type="natural earth",
                            showland=True,
                            landcolor="rgb(243, 243, 243)",
                            coastlinecolor="rgb(204, 204, 204)",
                            showocean=True,
                            oceancolor="rgb(230, 245, 255)",
                            showlakes=True,
                            lakecolor="rgb(230, 245, 255)",
                            showrivers=True,
                            rivercolor="rgb(230, 245, 255)"
                        )
                        
                        # Personnaliser les hover templates
                        fig.update_traces(
                            hovertemplate='<b>%{customdata[0]}</b><br>' +
                                          'Latitude: %{lat:.1f}°<br>' +
                                          'Longitude: %{lon:.1f}°<br>' +
                                          'Probabilité: %{customdata[1]:.3f}<br>' +
                                          'Observé: %{customdata[2]}<extra></extra>',
                            customdata=df_prediction_clean[['probability_class', 'probability', 'observed_presence']].values
                        )
                        
                        fig.update_layout(
                            title=dict(
                                text=f'🧬 Prédiction ML - Distribution de {category_name}',
                                x=0.5,
                                font=dict(size=16)
                            ),
                            width=1000,
                            height=700,
                            legend=dict(
                                title="Probabilité de Présence",
                                orientation="v",
                                yanchor="top",
                                y=1,
                                xanchor="left",
                                x=1.02
                            )
                        )
                        
                        fig.show()
                        
                        # Afficher un résumé des zones à haute probabilité
                        high_prob_cells = df_prediction_clean[df_prediction_clean['probability'] >= 0.6]
                        if len(high_prob_cells) > 0:
                            print(f"\n🎯 Zones à haute probabilité (≥60%):")
                            print(f"   Nombre de cellules: {len(high_prob_cells)}")
                            print(f"   Latitude: {high_prob_cells['center_lat'].min():.1f}° à {high_prob_cells['center_lat'].max():.1f}°")
                            print(f"   Longitude: {high_prob_cells['center_lon'].min():.1f}° à {high_prob_cells['center_lon'].max():.1f}°")
                        
                        print(f"\n🎉 Analyse ML terminée pour {category_name} !")
                        print(f"📁 Fichier généré: {output_filename}")
                        
                    else:
                        print(f"❌ Aucune cellule avec présence observée pour {category_name}")
                        
                else:
                    print(f"❌ Aucune variable environnementale disponible pour la prédiction")
            else:
                print(f"❌ Aucune donnée de grille disponible pour la prédiction")
        else:
            print(f"❌ Données insuffisantes pour {category_name}")
else:
    print("❌ Données agrégées non disponibles pour la prédiction ML")

print("\n" + "="*60)
print("💡 Pour analyser une autre catégorie, changez la variable 'category_name' et relancez cette cellule")
```

```python
# ==============================================================================
# PRÉDICTION MACHINE LEARNING PAR CATÉGORIE TAXONOMIQUE
# ==============================================================================
# Code général pour prédire la distribution d'une catégorie taxonomique
# Changez simplement la variable 'category_name' pour analyser différentes catégories

# ⚙️ CONFIGURATION - Changez cette ligne pour analyser différentes catégories
category_name = "Reptilia"  # Peut être changé pour: "Actinopterygii", "Aves", "Reptilia", etc.

print(f"🧬 Analyse ML pour la catégorie: {category_name}")
print("="*60)

if 'df_grid_groups' in locals() and df_grid_groups is not None:
    # Vérifier si la catégorie existe dans les données
    available_categories = df_grid_groups['group_analysis'].unique()
    print(f"📋 Catégories disponibles: {len(available_categories)}")
    print(f"🔍 Top 10 catégories: {available_categories[:10].tolist()}")
    
    if category_name not in available_categories:
        print(f"❌ Catégorie '{category_name}' non trouvée dans les données")
        print(f"💡 Catégories similaires disponibles:")
        similar_categories = [cat for cat in available_categories if category_name.lower() in cat.lower()]
        if similar_categories:
            for cat in similar_categories[:5]:
                print(f"   - {cat}")
        else:
            print(f"   Aucune catégorie similaire trouvée")
            print(f"   Essayez une de ces catégories: {available_categories[:5].tolist()}")
    else:
        print(f"✅ Catégorie '{category_name}' trouvée dans les données")
        
        # Filtrer les données pour la catégorie sélectionnée
        category_data = df_grid_groups[df_grid_groups['group_analysis'] == category_name].copy()
        print(f"📊 Données pour {category_name}:")
        print(f"   - Cellules avec observations: {len(category_data)}")
        print(f"   - Observations totales: {category_data['observation_count_sum'].sum():,}")
        print(f"   - Étendue géographique:")
        
        # Calculer l'étendue géographique
        lat_coords = category_data['grid_lat'] * 1.8 - 90 + 0.9
        lon_coords = category_data['grid_lon'] * 3.6 - 180 + 1.8
        print(f"     Latitude: {lat_coords.min():.1f}° à {lat_coords.max():.1f}°")
        print(f"     Longitude: {lon_coords.min():.1f}° à {lon_coords.max():.1f}°")
        
        # Variables environnementales disponibles
        env_columns = [col for col in df_grid_groups.columns if col.endswith('_mean_mean')]
        print(f"🌡️ Variables environnementales disponibles: {len(env_columns)}")
        
        if len(env_columns) > 0 and len(category_data) > 0:
            print(f"\n🤖 Création du modèle ML pour {category_name}...")
            
            # Préparer les données pour ML
            # Créer une grille complète de toutes les cellules possibles
            all_lat_indices = range(100)  # Grille 100x100
            all_lon_indices = range(100)
            
            ml_prediction_data = []
            
            for lat_idx in all_lat_indices:
                for lon_idx in all_lon_indices:
                    # Coordonnées géographiques réelles
                    center_lat = lat_idx * 1.8 - 90 + 0.9
                    center_lon = lon_idx * 3.6 - 180 + 1.8
                    
                    # Chercher les données environnementales pour cette cellule
                    cell_env_data = df_grid_groups[
                        (df_grid_groups['grid_lat'] == lat_idx) & 
                        (df_grid_groups['grid_lon'] == lon_idx)
                    ]
                    
                    if len(cell_env_data) > 0:
                        # Utiliser les données environnementales moyennes de cette cellule
                        row = {
                            'grid_lat': lat_idx,
                            'grid_lon': lon_idx,
                            'center_lat': center_lat,
                            'center_lon': center_lon
                        }
                        
                        # Variables environnementales
                        for env_col in env_columns:
                            if env_col in cell_env_data.columns:
                                row[env_col.replace('_mean_mean', '')] = cell_env_data[env_col].mean()
                        
                        # Vérifier si la catégorie est présente dans cette cellule
                        category_in_cell = cell_env_data[cell_env_data['group_analysis'] == category_name]
                        if len(category_in_cell) > 0:
                            row['observed_presence'] = 1
                            row['observation_count'] = category_in_cell['observation_count_sum'].sum()
                        else:
                            row['observed_presence'] = 0
                            row['observation_count'] = 0
                        
                        ml_prediction_data.append(row)
            
            # Créer le DataFrame pour les prédictions
            df_prediction = pd.DataFrame(ml_prediction_data)
            print(f"📊 Grille de prédiction créée: {len(df_prediction)} cellules")
            
            if len(df_prediction) > 0:
                # Calculer la probabilité de présence basée sur les variables environnementales
                # Utiliser une approche de similarité environnementale
                
                # Normaliser les variables environnementales
                env_feature_columns = [col for col in df_prediction.columns if col in [c.replace('_mean_mean', '') for c in env_columns]]
                
                if len(env_feature_columns) > 0:
                    print(f"🔬 Variables utilisées pour la prédiction: {env_feature_columns[:5]}{'...' if len(env_feature_columns) > 5 else ''}")
                    
                    # Remplir les valeurs manquantes
                    df_prediction_clean = df_prediction.copy()
                    for col in env_feature_columns:
                        if col in df_prediction_clean.columns:
                            df_prediction_clean[col] = df_prediction_clean[col].fillna(df_prediction_clean[col].mean())
                    
                    # Calculer les conditions environnementales moyennes où la catégorie est présente
                    present_cells = df_prediction_clean[df_prediction_clean['observed_presence'] == 1]
                    
                    if len(present_cells) > 0:
                        print(f"✅ Cellules avec présence observée: {len(present_cells)}")
                        
                        # Calculer les moyennes et écarts-types des conditions optimales
                        optimal_conditions = {}
                        for col in env_feature_columns:
                            if col in present_cells.columns:
                                optimal_conditions[col] = {
                                    'mean': present_cells[col].mean(),
                                    'std': present_cells[col].std() if present_cells[col].std() > 0 else 1.0
                                }
                        
                        # Calculer la probabilité de présence pour toutes les cellules
                        probabilities = []
                        
                        for _, row in df_prediction_clean.iterrows():
                            # Calculer la similarité environnementale
                            similarity_scores = []
                            
                            for col in env_feature_columns:
                                if col in optimal_conditions and col in row:
                                    # Distance normalisée par rapport aux conditions optimales
                                    optimal_mean = optimal_conditions[col]['mean']
                                    optimal_std = optimal_conditions[col]['std']
                                    
                                    if not pd.isna(row[col]) and optimal_std > 0:
                                        distance = abs(row[col] - optimal_mean) / optimal_std
                                        # Convertir la distance en score de similarité (0-1)
                                        similarity = np.exp(-distance**2 / 2)
                                        similarity_scores.append(similarity)
                            
                            # Probabilité moyenne basée sur toutes les variables
                            if len(similarity_scores) > 0:
                                probability = np.mean(similarity_scores)
                            else:
                                probability = 0.0
                            
                            probabilities.append(probability)
                        
                        df_prediction_clean['probability'] = probabilities
                        
                        # Classifier en 5 catégories de probabilité
                        def classify_probability(prob):
                            if prob >= 0.8:
                                return "Very Likely"
                            elif prob >= 0.6:
                                return "Likely"
                            elif prob >= 0.4:
                                return "Possible"
                            elif prob >= 0.2:
                                return "Unlikely"
                            else:
                                return "Very Unlikely"
                        
                        df_prediction_clean['probability_class'] = df_prediction_clean['probability'].apply(classify_probability)
                        
                        # Statistiques des prédictions
                        print(f"\n📈 Résultats de la prédiction pour {category_name}:")
                        prob_distribution = df_prediction_clean['probability_class'].value_counts()
                        for prob_class, count in prob_distribution.items():
                            percentage = (count / len(df_prediction_clean)) * 100
                            print(f"   {prob_class}: {count:,} cellules ({percentage:.1f}%)")
                        
                        print(f"\n📊 Statistiques des probabilités:")
                        print(f"   Probabilité moyenne: {df_prediction_clean['probability'].mean():.3f}")
                        print(f"   Probabilité médiane: {df_prediction_clean['probability'].median():.3f}")
                        print(f"   Probabilité max: {df_prediction_clean['probability'].max():.3f}")
                        print(f"   Probabilité min: {df_prediction_clean['probability'].min():.3f}")
                        
                        # Sauvegarder les résultats
                        output_filename = f'ml_prediction_{category_name.lower().replace(" ", "_")}.csv'
                        df_prediction_clean.to_csv(output_filename, index=False)
                        print(f"\n💾 Résultats sauvegardés: {output_filename}")
                        
                        # Créer la visualisation
                        print(f"\n🗺️ Création de la carte de prédiction...")
                        
                        # Définir les couleurs pour chaque classe de probabilité
                        color_map = {
                            "Very Likely": "#d73027",     # Rouge foncé
                            "Likely": "#fc8d59",          # Orange
                            "Possible": "#fee08b",        # Jaune
                            "Unlikely": "#e0f3f8",        # Bleu clair
                            "Very Unlikely": "#91bfdb"    # Bleu
                        }
                        
                        df_prediction_clean['color'] = df_prediction_clean['probability_class'].map(color_map)
                        
                        # Créer la carte avec Plotly
                        fig = px.scatter_geo(
                            df_prediction_clean,
                            lat='center_lat',
                            lon='center_lon',
                            color='probability_class',
                            color_discrete_map=color_map,
                            size='probability',
                            size_max=10,
                            hover_data=['probability', 'observed_presence'],
                            title=f'🧬 Prédiction ML - Distribution de {category_name}',
                            category_orders={"probability_class": ["Very Likely", "Likely", "Possible", "Unlikely", "Very Unlikely"]}
                        )
                        
                        # Configuration de la carte
                        fig.update_geos(
                            projection_type="natural earth",
                            showland=True,
                            landcolor="rgb(243, 243, 243)",
                            coastlinecolor="rgb(204, 204, 204)",
                            showocean=True,
                            oceancolor="rgb(230, 245, 255)",
                            showlakes=True,
                            lakecolor="rgb(230, 245, 255)",
                            showrivers=True,
                            rivercolor="rgb(230, 245, 255)"
                        )
                        
                        # Personnaliser les hover templates
                        fig.update_traces(
                            hovertemplate='<b>%{customdata[0]}</b><br>' +
                                          'Latitude: %{lat:.1f}°<br>' +
                                          'Longitude: %{lon:.1f}°<br>' +
                                          'Probabilité: %{customdata[1]:.3f}<br>' +
                                          'Observé: %{customdata[2]}<extra></extra>',
                            customdata=df_prediction_clean[['probability_class', 'probability', 'observed_presence']].values
                        )
                        
                        fig.update_layout(
                            title=dict(
                                text=f'🧬 Prédiction ML - Distribution de {category_name}',
                                x=0.5,
                                font=dict(size=16)
                            ),
                            width=1000,
                            height=700,
                            legend=dict(
                                title="Probabilité de Présence",
                                orientation="v",
                                yanchor="top",
                                y=1,
                                xanchor="left",
                                x=1.02
                            )
                        )
                        
                        fig.show()
                        
                        # Afficher un résumé des zones à haute probabilité
                        high_prob_cells = df_prediction_clean[df_prediction_clean['probability'] >= 0.6]
                        if len(high_prob_cells) > 0:
                            print(f"\n🎯 Zones à haute probabilité (≥60%):")
                            print(f"   Nombre de cellules: {len(high_prob_cells)}")
                            print(f"   Latitude: {high_prob_cells['center_lat'].min():.1f}° à {high_prob_cells['center_lat'].max():.1f}°")
                            print(f"   Longitude: {high_prob_cells['center_lon'].min():.1f}° à {high_prob_cells['center_lon'].max():.1f}°")
                        
                        print(f"\n🎉 Analyse ML terminée pour {category_name} !")
                        print(f"📁 Fichier généré: {output_filename}")
                        
                    else:
                        print(f"❌ Aucune cellule avec présence observée pour {category_name}")
                        
                else:
                    print(f"❌ Aucune variable environnementale disponible pour la prédiction")
            else:
                print(f"❌ Aucune donnée de grille disponible pour la prédiction")
        else:
            print(f"❌ Données insuffisantes pour {category_name}")
else:
    print("❌ Données agrégées non disponibles pour la prédiction ML")

print("\n" + "="*60)
print("💡 Pour analyser une autre catégorie, changez la variable 'category_name' et relancez cette cellule")
```
